# RazorRisk Hyperparameter Search

Leakage-aware 3-fold cross-validation on the synthetic RazorRisk dataset. The final held-out test split is not used for hyperparameter selection. XGBoost and GNN use user-level folds; the stacker is tuned from out-of-fold base-model predictions and is explicitly class-balanced.

## Selection policy

- Primary metric: mean CV PR-AUC
- Secondary metric: mean CV ROC-AUC
- XGBoost: user-level folds + fold-only merchant target encoding + fold-specific `scale_pos_weight`
- GNN: user-level folds and positive-class weighted loss
- Stacker: out-of-fold XGBoost/GNN predictions + `class_weight="balanced"`; only balanced candidates are considered
- Final test set: evaluated only after selection/retraining
